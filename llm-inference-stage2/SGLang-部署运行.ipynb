## 启动 SGLang Server

在 `ai-infra-lab` 目录下执行以下 bash 命令，通过 `--download-dir` 指定模型下载路径（会自行创建 models 文件夹）。

```bash
python3 -m sglang.launch_server \
    --model-path Qwen/Qwen3-0.6B \
    --download-dir . \
    --host 0.0.0.0 --port 30000
```

后续再次启动用如下命令即可：

```bash
python3 -m sglang.launch_server \
    --model-path ./models/Qwen--Qwen3-0.6B/snapshots/master \
    --host 0.0.0.0 --port 30000
```

如果设置了代理导致报错可先用如下命令取消报错（Linux）：

```bash
unset http_proxy https_proxy all_proxy
unset HTTP_PROXY HTTPS_PROXY ALL_PROXY

export NO_PROXY="localhost,127.0.0.1,0.0.0.0"
export no_proxy="localhost,127.0.0.1,0.0.0.0"
```

启动成功后验证：

```bash
curl http://localhost:30000/health
```

发送第一个对话请求：

```bash
curl -s http://localhost:30000/v1/chat/completions \
  -H "Content-Type: application/json" \
  -d '{
    "model": "./models/Qwen--Qwen3-0.6B/snapshots/master",
    "messages": [{"role": "user", "content": "hello!"}],
    "max_tokens": 512
  }'
```

Python 调用：

````python
import openai
import os
# 强制清除环境变量里的代理
os.environ["http_proxy"] = ""
os.environ["https_proxy"] = ""
os.environ["all_proxy"] = ""
os.environ["HTTP_PROXY"] = ""
os.environ["HTTPS_PROXY"] = ""

client = openai.Client(base_url="http://127.0.0.1:30000/v1", api_key="None")

response = client.chat.completions.create(
    model="../models/Qwen--Qwen3-0.6B/snapshots/master",
    messages=[{"role": "user", "content": "列举 3 个国家和它们的首都。"}],
    max_tokens=512,
)
print(response.choices[0].message.content)
````
